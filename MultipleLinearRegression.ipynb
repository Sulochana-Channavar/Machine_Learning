```python
import numpy as np
import pandas as pd

# =========================================================================
# STEP 1: LOAD YOUR OWN DATASET
# =========================================================================
# Replace 'your_dataset.csv' with the path to your actual CSV file!
# Example CSV structure:
# SquareFootage, Bedrooms, Bathrooms, Price
# 2104,          3,        2,         400
# 1600,          3,        1,         330

try:
    df = pd.read_csv(r"\Users\suloc\Downloads\housing_price.csv")
    
    # Separate your input features (X) and target output (y)
    # Change these column names to match the headers in your CSV!
    feature_columns = ['SquareFootage', 'Bedrooms', 'Bathrooms', 'Age']
    target_column = 'Price'
    
    X_raw = df[feature_columns].values.astype(float)
    y = df[target_column].values.reshape(-1, 1).astype(float)

except FileNotFoundError:
    print("⚠️ CSV file not found! Using generated sample data instead...\n")
    # Backup dummy data if file isn't created yet
    X_raw = np.array([[2104, 3], [1600, 3], [2400, 3], [1416, 2], [3000, 4]], dtype=float)
    y = np.array([400, 330, 369, 232, 540], dtype=float).reshape(-1, 1)

m = X_raw.shape[0]  # Total rows in your dataset

# =========================================================================
# STEP 2: FEATURE SCALING (Required for Gradient Descent)
# =========================================================================
# We calculate Mean (mu) and Standard Deviation (sigma) from YOUR dataset
mu = np.mean(X_raw, axis=0)
sigma = np.std(X_raw, axis=0)

# Normalize features: (x - mean) / std_dev
X_scaled = (X_raw - mu) / sigma

# Add column of 1s at index 0 for the bias term (theta_0)
X = np.hstack((np.ones((m, 1)), X_scaled))

# =========================================================================
# STEP 3: APPROACH 1 - GRADIENT DESCENT (Iterative)
# =========================================================================
def gradient_descent(X, y, alpha=0.01, iterations=1500):
    m, n = X.shape
    theta = np.zeros((n, 1))
    
    for i in range(iterations):
        predictions = np.dot(X, theta)
        errors = predictions - y
        gradients = (1 / m) * np.dot(X.T, errors)
        theta = theta - alpha * gradients
        
    return theta

theta_gd = gradient_descent(X, y, alpha=0.01, iterations=1500)

# =========================================================================
# STEP 4: APPROACH 2 - NORMAL EQUATION (Closed-Form Analytical)
# =========================================================================
def normal_equation(X, y):
    # theta = inv(X^T * X) * X^T * y
    X_transpose = X.T
    theta = np.linalg.inv(X_transpose.dot(X)).dot(X_transpose).dot(y)
    return theta

theta_ne = normal_equation(X, y)

# =========================================================================
# STEP 5: MAKE A PREDICTION ON NEW DATA
# =========================================================================
# Pass custom values for a new query sample! 
# (e.g., a house with 2200 SqFt and 3 Bedrooms)
my_new_sample = np.array([2200, 3, 2, 10], dtype=float)

# Scale using the EXACT same mu and sigma from your dataset!
my_sample_scaled = (my_new_sample - mu) / sigma
my_sample_input = np.append([1], my_sample_scaled).reshape(1, -1)

# Compute prediction
pred_gd = np.dot(my_sample_input, theta_gd)[0][0]
pred_ne = np.dot(my_sample_input, theta_ne)[0][0]

print("=" * 60)
print(f"Dataset Size Loaded: {m} rows")
print(f"Learned Weights (Gradient Descent):\n{theta_gd.ravel()}")
print(f"Learned Weights (Normal Equation):\n{theta_ne.ravel()}")
print("-" * 60)
print(f"🎯 Prediction for sample {my_new_sample}:")
print(f"   ► Gradient Descent Prediction : {pred_gd:.2f}")
print(f"   ► Normal Equation Prediction    : {pred_ne:.2f}")
print("=" * 60)
```
