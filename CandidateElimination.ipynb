```python
import numpy as np
import pandas as pd

# 1. Load your own custom CSV dataset
df = pd.read_csv(r'\Users\suloc\Downloads\training_data.csv') 

# 2. Extract features (concepts) and target labels dynamically
# Assumes your last column is the target (e.g., 'Yes'/'No' or 'Positive'/'Negative')
concepts = np.array(df.iloc[:, :-1])
target = np.array(df.iloc[:, -1])

# 3. Initialize boundaries based on your dataset's shape
S = concepts[0].copy()
G = [["?" for _ in range(len(S))] for _ in range(len(S))]

# 4. Loop through your dataset instances
for i, instance in enumerate(concepts):
    # NOTE: Change "Yes" to whatever your positive label is in your dataset (e.g., 1, "Pass", etc.)
    if target[i] == "Yes":  
        for x in range(len(S)):
            if instance[x] != S[x]:
                S[x] = '?'
                G[x][x] = '?'
                
    else:  # For Negative instances
        for x in range(len(S)):
            if instance[x] != S[x]:
                G[x][x] = S[x]
            else:
                G[x][x] = '?'

# 5. Clean up the General boundary
G_final = [row for row in G if row != ['?'] * len(S)]

print("Final Specific Boundary (S):", S)
print("Final General Boundary (G):", G_final)
```
