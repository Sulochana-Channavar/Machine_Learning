```python
import numpy as np


class DecisionTreeScratch:

  def __init__(self, max_depth=5, min_samples_split=2):
    self.max_depth = max_depth
    self.min_samples_split = min_samples_split
    self.tree = None

  def _gini_impurity(self, y):
    if len(y) == 0:
      return 0.0
    p = np.bincount(y) / len(y)
    return 1.0 - np.sum(p**2)

  def _information_gain(self, y, y_left, y_right):
    weight_left = len(y_left) / len(y)
    weight_right = len(y_right) / len(y)
    gain = (
        self._gini_impurity(y)
        - (weight_left * self._gini_impurity(y_left))
        - (weight_right * self._gini_impurity(y_right))
    )
    return gain

  def _best_split(self, X, y):
    best_gain = -1
    split_idx, split_threshold = None, None
    m, n = X.shape

    for feat_idx in range(n):
      thresholds = np.unique(X[:, feat_idx])
      for thresh in thresholds:
        left_mask = X[:, feat_idx] <= thresh
        right_mask = ~left_mask

        if np.sum(left_mask) == 0 or np.sum(right_mask) == 0:
          continue

        gain = self._information_gain(y, y[left_mask], y[right_mask])

        if gain > best_gain:
          best_gain = gain
          split_idx = feat_idx
          split_threshold = thresh

    return split_idx, split_threshold, best_gain


# =========================================================================
# DEMO EXECUTION
# =========================================================================
if __name__ == "__main__":
  # Features: [Age, Income_k$] -> Purchased Item (0 or 1)
  X = np.array(
      [[22, 25], [25, 32], [47, 70], [52, 110], [46, 62], [56, 120]]
  )
  y = np.array([0, 0, 1, 1, 1, 1])

  dt = DecisionTreeScratch()
  feat, thresh, gain = dt._best_split(X, y)

  print("=" * 55)
  print("  DECISION TREE ROOT SPLIT EVALUATION")
  print("=" * 55)
  print(f"Best Feature Index Split : Feature {feat}")
  print(f"Optimal Threshold Value  : <= {thresh}")
  print(f"Information Gain Yielded : {gain:.4f}")
  print("=" * 55)
```
